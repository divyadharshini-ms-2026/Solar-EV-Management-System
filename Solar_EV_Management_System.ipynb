```python
import numpy as np
import matplotlib.pyplot as plt

print("Solar EV Charging Station")
print("-------------------------")
print("Python setup successful!")

solar_panel_capacity = 5
battery_capacity = 10
ev_charger_power = 3.3

print("Solar Panel:", solar_panel_capacity, "kW")
print("Battery:", battery_capacity, "kWh")
print("EV Charger:", ev_charger_power, "kW")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Time in hours
time = np.array([6, 8, 10, 12, 14, 16, 18])

# Solar power generated in kW
solar_power = np.array([0.2, 1, 2.5, 4.5, 5, 3, 0.3])

# Display results
print("Time:", time)
print("Solar Power:", solar_power)

# Plot graph
plt.plot(time, solar_power, marker='o')
plt.xlabel("Time (Hours)")
plt.ylabel("Solar Power (kW)")
plt.title("Solar Power Generation")
plt.grid(True)
plt.show()
```

```python
energy = np.trapezoid(solar_power, time)
print("Total energy (kWh):", energy)
print("Peak power (kW):", max(solar_power))
```

```python
rate = 5 # change this to your electricity price per kWh

daily = energy * rate
monthly = daily * 30
yearly = daily * 365

print("Daily savings:", round(daily, 2))
print("Monthly savings:", round(monthly, 2))
print("Yearly savings:", round(yearly, 2))
```

Conclusion: The solar system's output rises from morning, peaks at 5 kW around 2 PM, and then drops toward evening. Total energy generated in one day (6 AM to 6 PM) is about 32.5 kWh. This shows that solar power is strongest in the midday hours. The 5 kW solar panel and 10 kWh battery can supply the 3.3 kW EV charger during peak sunlight hours. At a rate of 5 per kWh, the system saves about 162.5 per day, which is about 59,312 per year.
